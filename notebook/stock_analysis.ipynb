## Stock Market Analysis
this project analyzes stock market data using python.
it focuses on returns, volatility, and comparsion of multiple stocks.

## Data collection
In this section, This project uses 
**synthetic (simulated) financial data** 
generated to mimic real market behavior..

```python
import pandas as pd

df = pd.read_csv(
    "yahoo_like_2020_2025.csv",
    header=[0,1],
    index_col=0,
    parse_dates=True
)

df = df["MSFT"]

print(df.head())
```

 ## Ruterns Analysis
We calculate daily returns to measure stock performance.

```python

df['Daily_Return'] = df['Adj Close'].pct_change()
```

## Cumulative Return
This shows how an investment grows over time.

```python
df['Cumulative_Return'] = (1 + df['Daily_Return']).cumprod()
```

 ## Volatility Analysis
We measure risk using standard deviation of returns

```python
volatility = df['Daily_Return'].std()
print("Volatility:", volatility)
```

```python
annual_volatility = df['Daily_Return'].std() * (252 ** 0.5)
print("Annual Vilatility:", annual_volatility)
```

## Visualization
We visualize stock prices and returns. 

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(df['Adj Close'])
plt.title("Stock Price")
plt.xlabel("Dte")
plt.ylabel("Price")
plt.savefig("aapl_price.png")
plt.show()
```

## Daily Returns Analysis
In this section, we visualize the daily ruterns of each stock.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(df['Daily_Return'])
plt.title("Daily Return")
plt.savefig("aapl_dailyreturn.png")
plt.show()
```

## Multi-stock Comparsion
In this section, we compare multiple stocks to evaluate their performance.
this helps us understand which stock performs better and which on is riskier.

```python
import pandas as pd

tickers = ["AAPL", "MSFT", "TSLA"]
data = pd.read_csv(
    "yahoo_like_2020_2025.csv",
    header=[0,1],
    index_col=0,
    parse_dates=True
)

adj_close = data.xs("Adj Close", axis=1, level=1)

returns = adj_close.pct_change()

print(adj_close.head())
print(returns.head())
```

## Multi-stock Comparsion
In this section, we show multiple stocks to evaluate their performance.

```python
cumulative_returns = (1 + returns).cumprod()
cumulative_returns.plot(figsize=(10,5))
plt.title("Cumulative Return Comparison")
plt.savefig("cumulative.png")
plt.show()
```
